# Module 8 - Class 1: End-to-End ML Pipeline Design

**Problem: Uztelecom churn prediction**

```mermaid
flowchart LR
A[Billing, usage, support data] --> B[Data validation]
B --> C[Feature engineering]
C --> D[Train and track]
D --> E[Evaluate and approve]
E --> F[Flask API]
F --> G[Monitor drift and outcomes]
G --> C
```

## Components

- **Data source:** approved billing, usage, and support data is collected daily.
- **Validation:** required columns, types, ranges, and duplicates are checked before scoring.
- **Features:** tenure, service use, payment, and support-contact features are engineered.
- **Training:** Logistic Regression baseline and Random Forest challenger are tracked in MLflow.
- **Evaluation:** compare precision, recall, F1, calibration, and fairness.
- **Deployment:** an approved full pipeline is served through Flask.
- **Monitoring:** log latency, prediction distribution, data drift, and delayed outcomes.

## Risks

| Risk | Mitigation |
|---|---|
| Schema change | Versioned data contracts and automated validation |
| Data drift | Monitor distributions and F1; retrain after a threshold |
| Traffic spike | Rate limits, autoscaling, and batch queue |
| Privacy breach | Least privilege, encryption, and no raw PII in logs |

Logistic Regression is an interpretable binary-classification baseline. F1 is appropriate because churn is imbalanced and both missed churners and unnecessary offers are costly.
