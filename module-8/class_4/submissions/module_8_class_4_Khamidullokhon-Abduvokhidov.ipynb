# Module 8 - Class 4: Containerizing the ML Application

The deployable Docker project is in `module_8_class_4_container/`. First run the Class 3 training cell and copy `model.pkl` into this folder. Then run:

```bash
docker build -t churn-api:v1 module_8_class_4_container
docker run --rm -p 5000:5000 churn-api:v1
curl http://localhost:5000/health
```

## Portability notes

The image uses `python:3.10-slim`, pins dependencies, installs them before copying application code for cache efficiency, and binds Flask to `0.0.0.0`. A clean rebuild should work wherever Docker is installed, provided `model.pkl` is included in the build context.
