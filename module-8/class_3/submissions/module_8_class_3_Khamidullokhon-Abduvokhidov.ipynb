# Module 8 - Class 3: Building a Prediction API with Flask

This notebook trains and saves a breast-cancer pipeline, then provides a Flask API implementation with `/health` and `/predict`.

```
# Train and save one complete preprocessing-and-model pipeline.
from sklearn.datasets import load_breast_cancer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
import joblib
data = load_breast_cancer()
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))
model.fit(data.data, data.target)
joblib.dump(model, 'model.pkl')
print('Saved model with', data.data.shape[1], 'features.')
```

```
# Test the Flask app after saving app.py from the project folder.
from module_8_class_3_app import app
client = app.test_client()
print(client.get('/health').get_json())
print(client.post('/predict', json={'features': data.data[0].tolist()}).get_json())
print(client.post('/predict', json={'features': [1, 2]}).status_code)
print(client.post('/predict', json={'features': ['bad'] * 30}).status_code)
```
