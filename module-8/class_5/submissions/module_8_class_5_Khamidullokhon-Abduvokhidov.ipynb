# Module 8 - Class 5: Capstone Project Development

## Project: Uztelecom Customer Churn Prediction

**Objective:** predict whether an active customer will churn in the next 30 days. Success means recall at least 75% while tracking precision and F1 so the retention team does not receive excessive false alarms. ML is appropriate because churn results from combinations of tenure, service use, payment history, and support interactions that are difficult to score manually.

## Data and feature plan

Use a real dataset with 1,000+ rows, such as Telco Customer Churn. Clean numeric charge fields, remove identifiers from features, and split train/validation/test before fitting preprocessing. Engineered features: (1) `average_monthly_spend = TotalCharges / max(tenure, 1)`, representing spend intensity; (2) `is_new_customer = tenure < 6`, representing early churn risk; and (3) `support_risk`, combining online security, backup, and tech support status.

Compare Logistic Regression and Random Forest across at least three parameter settings each. Log parameters, F1, recall, precision, and AUC to MLflow. Select the model using held-out validation performance, calibration, latency, and interpretability.
