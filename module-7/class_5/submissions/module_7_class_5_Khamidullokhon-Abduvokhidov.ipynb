```
# Install YOLO, run it on five supplied images, and save annotated outputs.
!pip -q install ultralytics
from ultralytics import YOLO
import matplotlib.pyplot as plt,pandas as pd
model=YOLO('yolov8n.pt')
sources=['street.jpg','indoor.jpg','crowded.jpg','low_light.jpg','choice.jpg']
rows=[]
for source in sources:
  r=model(source)[0]; boxes=r.boxes; classes=[r.names[int(c)] for c in boxes.cls]; conf=boxes.conf.cpu().numpy()
  plt.figure(figsize=(10,6)); plt.imshow(r.plot()[...,::-1]); plt.axis('off'); plt.title(source); plt.show()
  rows.append([source,len(classes),sorted(set(classes)),f'{conf.min():.2f}-{conf.max():.2f}' if len(conf) else 'none'])
display(pd.DataFrame(rows,columns=['Image','Detected count','Classes','Confidence range']))
```
