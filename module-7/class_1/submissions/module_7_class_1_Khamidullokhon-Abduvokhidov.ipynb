Manual result: lowercase, no punctuation, tokens, then stopword removal and stemming. Removing `not` changes `not great` into `great`, which can reverse sentiment; preserve negations in a sentiment-specific stopword list.

```
# Verify manual preprocessing with NLTK.
import nltk,string
nltk.download('stopwords'); nltk.download('punkt'); nltk.download('punkt_tab')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
text="Uztelecom's customer service is NOT great!!! I've been waiting for 3 hours... Will they ever fix the network??"
lower=text.lower(); clean=lower.translate(str.maketrans('', '', string.punctuation)); tokens=word_tokenize(clean); filtered=[w for w in tokens if w not in stopwords.words('english')]; stemmed=[PorterStemmer().stem(w) for w in filtered]
print(lower, clean, tokens, filtered, stemmed, sep='\n')
```
