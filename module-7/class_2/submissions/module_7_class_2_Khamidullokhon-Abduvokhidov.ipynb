```
# Build and compare Bag-of-Words and TF-IDF telecom matrices.
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
corpus=['Internet speed is slow today','Customer support solved my problem','Mobile data package is affordable','Internet connection drops at night','Helpful support improves customer satisfaction']
cv=CountVectorizer(); bow=cv.fit_transform(corpus); display(pd.DataFrame(bow.toarray(),columns=cv.get_feature_names_out()))
tv=TfidfVectorizer(); tfidf=tv.fit_transform(corpus); display(pd.DataFrame(tfidf.toarray().round(3),columns=tv.get_feature_names_out()))
```

Words unique to one sentence receive higher TF-IDF weights, while repeated terms such as internet or customer are downweighted. BoW keeps raw counts; TF-IDF makes common words less influential.
