```
# Compare pretrained DistilBERT predictions to the TF-IDF baseline.
!pip -q install transformers
import time,pandas as pd,numpy as np
from transformers import pipeline
from sklearn.metrics import classification_report,accuracy_score,f1_score
sample=te.sample(min(1000,len(te)),random_state=42); start=time.time(); lr=m.predict(v.transform(sample.clean)); lr_time=(time.time()-start)/len(sample)
pipe=pipeline('sentiment-analysis',model='distilbert-base-uncased-finetuned-sst-2-english'); start=time.time(); out=pipe(sample.Text.tolist(),truncation=True); hf=np.array([x['label']=='POSITIVE' for x in out],int); hf_time=(time.time()-start)/len(sample)
display(pd.DataFrame([['TF-IDF + LogReg',accuracy_score(sample.y,lr),f1_score(sample.y,lr),lr_time],['DistilBERT',accuracy_score(sample.y,hf),f1_score(sample.y,hf),hf_time]],columns=['Model','Accuracy','F1','Seconds/Review']))
display(sample.loc[lr!=hf,['Text','y']].assign(LogReg=lr[lr!=hf],DistilBERT=hf[lr!=hf]).head(10))
```

## Production recommendation
Use TF-IDF plus Logistic Regression when low latency, low cost, and simple maintenance matter and accuracy is competitive. Use DistilBERT when its accuracy gain on representative product reviews justifies GPU or managed-inference cost. Validate both on the target domain because the pretrained transformer was trained on movie reviews.
