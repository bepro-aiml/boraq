```python
# Define and label the six points for the manual K-Means exercise.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
points = np.array([[1,2],[2,1],[1.5,1.5],[8,8],[9,7],[8.5,8.5]])
plt.scatter(points[:,0], points[:,1])
for x,y in points: plt.annotate(f'({x}, {y})', (x,y))
plt.xlim(0,10); plt.ylim(0,10); plt.grid(); plt.title('Six Data Points'); plt.show()
```

```python
# Show two assign-update iterations using the stated starting centroids.
centroids = np.array([[1,2],[9,7]], dtype=float)
for iteration in [1,2]:
    distances = np.linalg.norm(points[:,None,:] - centroids[None,:,:], axis=2)
    labels = distances.argmin(axis=1)
    print(f'Iteration {iteration}')
    display(pd.DataFrame(distances.round(2), columns=['Distance to C1','Distance to C2']).assign(Cluster=labels+1))
    centroids = np.array([points[labels == k].mean(axis=0) for k in range(2)])
    print('Updated centroids:', centroids)
manual_labels = labels.copy()
```

```python
# Verify manual assignments and plot the final sklearn clusters.
kmeans = KMeans(n_clusters=2, init=np.array([[1,2],[9,7]]), n_init=1, random_state=42).fit(points)
print('Manual labels:', manual_labels); print('sklearn labels:', kmeans.labels_)
plt.scatter(points[:,0], points[:,1], c=kmeans.labels_, cmap='viridis', s=80, label='Points')
plt.scatter(kmeans.cluster_centers_[:,0], kmeans.cluster_centers_[:,1], marker='X', s=250, c='red', label='Centroids')
plt.xlabel('x'); plt.ylabel('y'); plt.title('K-Means Clusters'); plt.legend(); plt.show()
```

## Reflection
The manual calculation and sklearn result match up to arbitrary cluster-label numbering. The two groups converge to the three low-coordinate points and the three high-coordinate points after the assign-update steps.
