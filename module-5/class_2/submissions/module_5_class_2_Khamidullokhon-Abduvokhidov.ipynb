```python
# Generate and display two moon-shaped clusters.
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
X, y = make_moons(n_samples=300, noise=0.1, random_state=42)
plt.scatter(X[:,0], X[:,1]); plt.title('Moon-Shaped Data'); plt.show()
```

```python
# Compare K-Means, density-based DBSCAN, and agglomerative clustering.
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10).fit(X)
dbscan = DBSCAN(eps=0.2, min_samples=5).fit(X)
agg = AgglomerativeClustering(n_clusters=2).fit(X)
fig, axes = plt.subplots(1,3,figsize=(15,4))
for ax, labels, title in zip(axes, [kmeans.labels_, dbscan.labels_, agg.labels_], ['K-Means','DBSCAN','Agglomerative']):
    ax.scatter(X[:,0], X[:,1], c=labels, cmap='viridis'); ax.set_title(title)
plt.show()
```

## Explanation
K-Means partitions points around centroids and therefore prefers round, convex groups; it cuts across the crescent shapes. DBSCAN links nearby dense points, so it can follow each curved crescent and label isolated points as noise. Agglomerative clustering builds groups by merging nearby points and may capture some curved structure, depending on its linkage method.
