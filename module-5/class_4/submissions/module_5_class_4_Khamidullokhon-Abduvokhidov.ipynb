```python
# Load and standardize Mall Customers features, then obtain cluster labels.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
df = pd.read_csv('Mall_Customers.csv')
features = df.select_dtypes(include=np.number).drop(columns=['CustomerID'], errors='ignore')
X_scaled = StandardScaler().fit_transform(features)
labels = KMeans(n_clusters=5, random_state=42, n_init=10).fit_predict(X_scaled)
pca_points = PCA(n_components=2).fit_transform(X_scaled)
```

```python
# Run t-SNE and UMAP on the same scaled customers.
tsne_points = TSNE(n_components=2, perplexity=30, random_state=42).fit_transform(X_scaled)
try:
    import umap
except ImportError:
    !pip -q install umap-learn
    import umap
umap_points = umap.UMAP(n_components=2, random_state=42).fit_transform(X_scaled)
```

```python
# Compare the three two-dimensional cluster visualizations side by side.
fig, axes = plt.subplots(1,3,figsize=(18,5))
for ax, data, title in zip(axes, [pca_points, tsne_points, umap_points], ['PCA','t-SNE','UMAP']):
    ax.scatter(data[:,0], data[:,1], c=labels, cmap='viridis')
    ax.set_title(title)
plt.show()
```

## Method comparison
PCA is fast, deterministic, and interpretable because each axis is a linear combination of features, but it may miss nonlinear structure. t-SNE often makes local groups visually clear, though global distances should not be over-interpreted. UMAP generally preserves local structure efficiently and can retain more global organization than t-SNE. I would use PCA for explanation and preprocessing, and t-SNE or UMAP for exploratory visualization of possible segments.
