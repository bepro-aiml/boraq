```python
# Load Mall Customers data, select numeric customer features, and assign K-Means labels.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
df = pd.read_csv('Mall_Customers.csv')
features = df.select_dtypes(include=np.number).drop(columns=['CustomerID'], errors='ignore')
X_scaled = StandardScaler().fit_transform(features)
labels = KMeans(n_clusters=5, random_state=42, n_init=10).fit_predict(X_scaled)
```

```python
# Plot explained variance and report components needed for 90% and 95%.
pca_all = PCA().fit(X_scaled)
cumulative = np.cumsum(pca_all.explained_variance_ratio_)
plt.bar(range(1,len(cumulative)+1), pca_all.explained_variance_ratio_); plt.xlabel('Principal Component'); plt.ylabel('Explained Variance Ratio'); plt.title('PCA Scree Plot'); plt.show()
print('Components for 90%:', np.argmax(cumulative >= .90) + 1)
print('Components for 95%:', np.argmax(cumulative >= .95) + 1)
```

```python
# Reduce data to two components and interpret their feature loadings.
pca = PCA(n_components=2); pcs = pca.fit_transform(X_scaled)
plt.scatter(pcs[:,0], pcs[:,1], c=labels, cmap='viridis'); plt.xlabel('PC1'); plt.ylabel('PC2'); plt.title('PCA Customer Segments'); plt.show()
loadings = pd.DataFrame(pca.components_, columns=features.columns, index=['PC1','PC2'])
display(loadings)
print('The largest absolute loadings identify the original features that define each component.')
```
