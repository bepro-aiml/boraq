# Module 1 - Class 5: Research-Driven ML Problem Framing

## Scenario: Fraud detection for Payme / Click

**Business objective:** reduce financial loss while keeping legitimate mobile payments fast and minimally disrupted.

**Target variable:** whether a completed transaction is confirmed fraud within a defined investigation window, for example 30 days.

**ML problem type:** a hybrid of supervised binary classification and anomaly detection. Confirmed fraud labels train the classifier; anomaly detection helps find new attack patterns.

**Input features:**

1. Amount - unusual amounts indicate risk.
2. Transaction timestamp - captures unusual times.
3. Transaction type - risk differs across transfer and cash-out types.
4. Account age - new accounts may need different controls.
5. Device consistency - new-device events can be suspicious.
6. Recent transaction count - catches velocity attacks.
7. Recent amount total - catches rapid spending.
8. Time since last transaction - models account behavior.
9. Receiver risk history - identifies repeat destinations.
10. Approximate distance from usual region - catches impossible travel.
11. Authentication outcome - captures failed or weak verification signals.
12. Prior chargeback/fraud feedback - supplies learning signal.

**Evaluation:** optimize recall at a minimum precision chosen with the security team; report PR-AUC, F1, operational latency, and false-block rate. Fraud is likely far below 1% of transactions, so class weighting, time-aware validation, and delayed-label handling are required.

**Ethics and oversight:** location must never proxy ethnicity or social status. A low-confidence flag should trigger step-up authentication or analyst review, not automatic denial. A human must override any block, resolve appeals, and review severe or novel cases.

**Data availability in Uzbekistan:** transaction logs and authentication events would come from the payment platform, while labels come from investigations and chargebacks. The difficult part is safe access to sensitive financial data and reliable, timely labels; encryption, access control, retention limits, and audit logs are essential.

## Personal reflection

The hardest part of framing this problem was defining fraud and deciding how to handle false positives, because an incorrect block harms a legitimate customer immediately. I also found that the model cannot be designed separately from its review and appeal process. The rare positive class makes accuracy almost meaningless. Local data access and privacy controls are as important as the algorithm. The final design must treat the model as a risk signal, not an autonomous judge.
