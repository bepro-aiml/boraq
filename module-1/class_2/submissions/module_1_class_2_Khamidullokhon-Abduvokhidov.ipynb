# Module 1 - Class 2: Probability in Everyday Life

## Three examples

1. **Weather forecasting:** a forecast estimates the probability of rain tomorrow from satellite images, pressure, humidity, and historical weather patterns. Certainty is impossible because atmospheric measurements are incomplete and small changes can affect local rain.

2. **Spam filtering:** an email system estimates whether a message is spam using wording, sender behavior, links, and prior labels. It cannot be perfectly certain because attackers constantly change tactics and legitimate messages can resemble spam.

3. **Credit-card fraud alerts:** a payment provider estimates the chance that a transaction is fraudulent from amount, device, location, time, and account history. A legitimate traveler can look unusual and a fraudster can imitate normal behavior, so the decision remains probabilistic.

## Why ML cannot be 100% accurate

ML models work with imperfect observations of a changing world, not complete knowledge. Data can contain recording errors, missing variables, and biased samples. Even a well-trained model faces noise: two customers with identical recorded features can make different decisions for unrecorded reasons. The bias-variance tradeoff means a model that is too simple misses real patterns, while an overly complex one memorizes training noise. Distribution shift also matters: a spam filter trained on older email may miss new tactics, and a demand model may fail during unusual holidays. Evaluation scores estimate performance on a sample, not every future case. For these reasons, a probability and calibrated uncertainty are more honest than a claim of perfect accuracy.
