## Reflections

**Hardest scenarios:** Dynamic pricing, fraud, and Discover Weekly can be framed differently depending on implementation. I chose reinforcement learning for dynamic pricing because prices are actions with delayed reward, anomaly detection for incomplete fraud labels, and collaborative filtering for recommendations based on similarity.

**Multiple ML types:** One business system can use several approaches. A churn project can first cluster customers to find behavior segments, then train supervised churn models per segment, while a retention-offer policy could be optimized with reinforcement learning.

**Most useful for Uzbek telecom:** Supervised classification is immediately valuable because telecom has customer records and business outcomes such as cancellation, payment default, and support escalation. It can support retention and fraud workflows, but humans should review high-impact decisions.
