# Module 1 - Class 3: AI in Uzbekistan

| Application | Sector | Likely ML type | Data used |
|---|---|---|---|
| AI Assistant on my.gov.uz | Government services | NLP / retrieval-augmented chatbot | Service descriptions, regulations, user queries |
| Uzum Avto AI moderation | E-commerce / classifieds | Supervised classification and computer vision | Listing text, photos, duplicate signals, user reports |
| G'azna AI ChatBOT in UzASBO | Public finance | NLP assistant | Procurement rules, legal documents, user questions |

**my.gov.uz AI Assistant:** The official portal says its AI Assistant consults users about electronic public services and answers inquiries. This is most plausibly an NLP retrieval-and-generation system: it needs a curated knowledge base of approved service information and query/response logs. The portal appropriately states that generated answers are advisory, which is important where official information may change.

**Uzum Avto moderation:** Uzum says its AI-powered moderation system blocks duplicate, fake, and inaccurate vehicle listings. The technical details are not published, so supervised classification is an informed inference: historical listing decisions, text, photos, price, and duplicate patterns could train such a system. Human appeals and sampling should check for mistaken takedowns.

**G'azna AI:** The Treasury Service Committee reports that G'azna AI gives procurement guidance and is trained on current regulatory legal acts. This fits an NLP knowledge assistant grounded in regulations and open data. It should cite the relevant rule and route uncertain or high-impact questions to a human official.
