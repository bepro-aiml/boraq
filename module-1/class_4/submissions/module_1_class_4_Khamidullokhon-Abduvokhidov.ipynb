# Module 1 - Class 4: ITES Case Study - Mobile Payment Fraud Detection

## 1. Problem type

This is binary supervised classification: use confirmed historical fraud cases to predict whether an incoming transaction is fraudulent. It is difficult because fraud is rare, labels can arrive late, and fraud patterns change. An anomaly-detection score can supplement the classifier for novel tactics.

## 2. Data requirements

1. Transaction ID - joins feedback and investigation outcome.
2. Amount - unusually large or unusual amounts can raise risk.
3. Timestamp - supports velocity and unusual-time features.
4. Sender account age - new accounts may require different risk treatment.
5. Receiver account history - helps detect risky destinations.
6. Transaction type - transfers, payments, and cash-outs have different patterns.
7. Device identifier - detects new or rapidly changing devices.
8. Approximate location/region - supports impossible-travel checks, used carefully.
9. Recent transaction count and amount - detects velocity spikes.
10. Prior confirmed fraud/chargeback outcome - provides the training label.

## 3. Evaluation metric

I would use recall at a fixed, operationally acceptable precision, and report precision-recall AUC and F1. High recall protects customers by finding more fraud, but precision matters because false blocks interrupt legitimate payments. Accuracy is misleading because a model predicting every transaction as normal could look excellent when fraud is rare.

## 4. Bias and fairness

A model can over-flag customers from regions, languages, or income groups that were overrepresented in past investigations. Location should only be used for behavior-based signals such as a deviation from the same customer's normal region, not as a proxy for protected identity. Audit false-positive and false-negative rates by meaningful groups, investigate disparities, and provide a fast appeal path.

## 5. Privacy

Transaction, device, location, and account-history data reveal sensitive financial behavior. A breach could enable targeted scams or identity theft, so data should be minimized, encrypted, access-controlled, retained only as long as needed, and separated from direct identifiers where possible.

## 6. Recommendation

Deploy only with guardrails: real-time risk scores should trigger step-up verification or human review rather than permanent automatic blocks for borderline cases. Monitor drift, provide customer remediation, audit group-level error rates, and document every high-impact decision. Mastercard's fraud products demonstrate the real-world need for real-time risk decisions, but a local system still needs local validation.

## Reflections

Stricter privacy law would increase requirements for lawful basis, minimization, explanation, deletion, and auditability, but the ethical concerns already apply. A useful fraud system can use less personal data, but it will lose some contextual detection power; privacy-preserving aggregation is a worthwhile compromise. Responsibility belongs to the deploying company and accountable humans, not an algorithm; developers, data owners, and operational teams each share duties.
