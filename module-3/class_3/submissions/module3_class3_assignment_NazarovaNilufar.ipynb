```python
import pandas as pd #uploading pandas library
url = 'https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv'
df = pd.read_csv(url)#opening the file
```

```python
obj_cols = df.select_dtypes(include='object').columns.tolist() # it is categorical encoding, select_dtypes(include='object' , 'int') this selects only columns that you want to work with, .columngives the names of columns .tolist makes the indexing python list
print('Total object columns:', len(obj_cols)) #it gives the number of selected columns 'object'columns basically
print(obj_cols) #it shows all the columns we selected
```

```python
df['gender'].value_counts() #this shows the unique values of gender column, value_counts() is for this service
```

```python
dummies = pd.get_dummies(df['InternetService'], prefix='Internet') #it is one=hot encoding, dummies refers to the one-hot encoding, this code selects the column InternetService, makes one hot encoding, prefix show the new columnsare from that internet column
print('Shape:', dummies.shape) #shows the number of rows and columns of Internet Service column,(after encoding)
print(dummies) #this shows all the results of one=hot encoding
```

```python
dummies2 = pd.get_dummies(df['InternetService'], prefix='Internet', drop_first=True) #same thing happening like the previous cell, but in that one, we are removing the first column of new column we created by encoding
print('With drop_first:', dummies2.shape) #shows the number of rows and columns
dummies2.head() #shows the first 5rows of data
```

```python
from sklearn.preprocessing import OneHotEncoder
# Import OneHotEncoder to convert categorical values into binary columns.

ohe = OneHotEncoder(sparse_output=False)
# Create the encoder and return the result as a regular NumPy array.

encoded = ohe.fit_transform(df[['TechSupport']])
# Learn the unique categories and convert the column into one-hot encoded values.

encoded_df = pd.DataFrame(
    encoded,
    columns=ohe.get_feature_names_out(['TechSupport'])
)
# Convert the encoded array into a DataFrame with descriptive column names.

encoded_df.head()
# Display the first five rows of the encoded DataFrame.
```

```python
multi = pd.get_dummies(df[['gender', 'InternetService', 'TechSupport', 'Contract']], drop_first=True) #now, we are doing one =hot encoding with multiple columns, and dropping the first column of all thosecolumns
print('Shape:', multi.shape) #shows number of columns and rows of data after encoding
multi.head() #showing the first 5 rows
```
