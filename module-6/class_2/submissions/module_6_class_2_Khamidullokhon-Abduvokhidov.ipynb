```python
# Load normalized MNIST data and define one reusable dense architecture.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD, Adam
from tensorflow.keras.datasets import mnist
(X_train,y_train),(X_test,y_test)=mnist.load_data()
X_train=X_train.astype('float32')/255
X_test=X_test.astype('float32')/255
def build_model(): return Sequential([Flatten(input_shape=(28,28)),Dense(256,activation='relu'),Dense(128,activation='relu'),Dense(10,activation='softmax')])
def train(optimizer):
    model=build_model()
    model.compile(optimizer=optimizer,loss='sparse_categorical_crossentropy',metrics=['accuracy'])
    start=time.time()
    history=model.fit(X_train,y_train,epochs=20,batch_size=128,validation_split=.2,verbose=0)
    return model,history,time.time()-start
```

```python
# Train a newly initialized model with each optimizer and compare results.
runs={'SGD':train(SGD(.01)),'SGD + momentum':train(SGD(.01,momentum=.9)),'Adam':train(Adam(.001))}
rows=[]
for name,(model,h,elapsed) in runs.items():
    loss,acc=model.evaluate(X_test,y_test,verbose=0)
    rows.append([name,acc,elapsed])
display(pd.DataFrame(rows,columns=['Optimizer','Test Accuracy','Training Time']))
for name,(_,h,_) in runs.items(): plt.plot(h.history['loss'],label=name)
plt.legend(); plt.title('Training Loss'); plt.show()
for name,(_,h,_) in runs.items(): plt.plot(h.history['val_accuracy'],label=name)
plt.legend(); plt.title('Validation Accuracy'); plt.show()
```
