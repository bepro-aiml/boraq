```python
# Load MNIST and define baseline and dropout CNN architectures.
import numpy as np, matplotlib.pyplot as plt, time, tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D,MaxPooling2D,Flatten,Dense,Dropout
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
(X_train,y_train),(X_test,y_test)=mnist.load_data(); X_train=X_train.reshape(-1,28,28,1)/255.; X_test=X_test.reshape(-1,28,28,1)/255.; y_train=to_categorical(y_train); y_test=to_categorical(y_test)
def cnn(dropout=False):
  layers=[Conv2D(32,3,activation='relu',input_shape=(28,28,1)),MaxPooling2D(2)]
  if dropout: layers += [Dropout(.25)]
  layers += [Conv2D(64,3,activation='relu'),MaxPooling2D(2)]
  if dropout: layers += [Dropout(.25)]
  layers += [Flatten(),Dense(128,activation='relu')]
  if dropout: layers += [Dropout(.5)]
  layers += [Dense(10,activation='softmax')]
  m=Sequential(layers); m.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy']); return m
```

```python
# Train both models from scratch and compare learning curves.
baseline,regularized=cnn(),cnn(True); baseline.summary(); regularized.summary()
h1=baseline.fit(X_train,y_train,epochs=10,batch_size=32,validation_split=.2); h2=regularized.fit(X_train,y_train,epochs=10,batch_size=32,validation_split=.2)
for h,label in [(h1,'Baseline train'),(h1,'Baseline val'),(h2,'Dropout train'),(h2,'Dropout val')]: plt.plot(h.history['accuracy'] if 'train' in label else h.history['val_accuracy'],label=label)
plt.legend(); plt.title('Accuracy Comparison'); plt.show()
print('Baseline gap:',h1.history['accuracy'][-1]-h1.history['val_accuracy'][-1]); print('Dropout gap:',h2.history['accuracy'][-1]-h2.history['val_accuracy'][-1])
```

## Analysis
The model with the larger training-versus-validation accuracy gap is overfitting more. Dropout often lowers training accuracy because units are randomly disabled during training, but this can improve generalization and reduce the gap. Compare the final validation accuracies and gaps printed above to decide which model generalizes better in this run.
