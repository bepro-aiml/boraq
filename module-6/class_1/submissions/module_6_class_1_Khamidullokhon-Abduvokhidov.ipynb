## Analysis

A network with zero hidden layers can only draw a linear decision boundary, so it cannot separate the inside and outside of a circle well. Adding a hidden layer allows the network to combine nonlinear ReLU features into a more flexible boundary. Two neurons may capture only a rough approximation, while four neurons can create more pieces of the circular boundary. Adding a second hidden layer increases the model's capacity to compose features, which can make the boundary smoother and more accurate. More layers and neurons can learn richer patterns, but they can also make training slower and raise the risk of overfitting on more difficult data. Keeping the learning rate and activation fixed makes the comparison about architecture rather than training settings.
