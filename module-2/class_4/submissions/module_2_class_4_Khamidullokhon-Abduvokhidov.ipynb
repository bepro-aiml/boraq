### Step 0 — Load the data

```python
# Import pandas and load the Telco customer dataset
import pandas as pd

# Using a local copy with the same schema as the original Telco churn dataset
# (original assignment source: IBM Telco churn CSV on GitHub — offline here, so loading locally)
df = pd.read_csv('/home/claude/data/telco_churn_synthetic.csv')
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce').fillna(0)
print('Shape:', df.shape)
```

### Step 1 — See column info

```python
# Print the dtypes and non-null counts of every column
df.info()
```

### Step 2 — Statistics for numeric columns

```python
# Show min/max/mean/quartiles for every numeric column
df.describe().round(2)
```

### Step 3 — Count unique values per column

```python
# Show how many unique values are in each column
df.nunique()
```

### Step 4 — Filter — customers paying more than $80/month

```python
# Keep only rows where MonthlyCharges is above 80
high_pay = df[df['MonthlyCharges'] > 80]
print('High-paying customers:', len(high_pay))
high_pay.head()
```

### Step 5 — Sort — top 10 by tenure

```python
# Sort customers by tenure (longest first) and grab the top 10
top10 = df.sort_values('tenure', ascending=False).head(10)
top10[['customerID', 'tenure', 'Contract', 'MonthlyCharges']]
```

### Step 6 — Groupby — avg MonthlyCharges per PaymentMethod

```python
# Group by PaymentMethod and compute the average MonthlyCharges for each group
df.groupby('PaymentMethod')['MonthlyCharges'].mean().round(2)
```

### Step 7 — Groupby with two stats

```python
# Group by Churn and compute both the mean and the count of TotalCharges
df.groupby('Churn').agg({'TotalCharges': ['mean', 'count']}).round(2)
```

### Step 8 — Crosstab — Contract vs Churn

```python
# Cross-tabulate Contract type against Churn to see churn counts per contract
pd.crosstab(df['Contract'], df['Churn'])
```
